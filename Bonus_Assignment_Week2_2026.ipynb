## Hospital Care System

A hospital wants to modernize how it tracks daily patient care. Right now, everything is written on paper charts, and it's becoming impossible to keep up with who did what for which patient.

*Your job is to build a simple digital system that models a patient's treatment plan for a day.*

Every patient is assigned a team of care providers. This might include a doctor doing rounds, a nurse checking vitals, or a specialist brought in for a particular condition. Each type of provider interacts with the patient differently: a doctor examines and diagnoses, a nurse checks vitals, a specialist runs a consult in their field of expertise. Of course, in your code you only need to keep track of what is going on, so make sure that each care provider is logging what they have done for the patient (that the is only expected functionality).

On top of that, each patient may be undergoing one or more treatments as part of their care plan. Some treatments are medications, which need to be tracked carefully because there is a limited stock of each drug, and administering a dose should reduce what's left, with the system refusing to give more than what's available. Other treatments are surgeries, which don't get "used up" the way medication does, but do need to log how long they'll occupy an operating room. Still others are physical therapy sessions, which come with a fixed number of sessions remaining, counting down each time one is completed.

At the end of each day, the hospital wants to run a patient's entire care plan in one go: every assigned provider should do their part, and every treatment should be administered. The patient's chart should end up with a clear, readable log of everything that happened to them that day. You should be able to add care providers and treatments for a patient.

**Task 1: Design!** What classes would you like to use? What are the relationships between classes? Make a UML diagram (pen and paper is fine, add a picture to your submission).



**Task 2: Implement** the code and illustrate the functionally by running all the visits and treatments a patient would receive in one day.

**Example:** The output of executing a care plan for one day could look like this:

--- Care for Alex ---

 * Alex was examined and diagnosed by Dr. Kim
 * Alex had vitals checked by Nurse Rios
 * Alex had a cardiology consult with Dr. Chen
 * Alex was given 500mg of Amoxicillin (700mg remaining in stock)
 * Alex underwent Appendectomy surgery, booked for 90 minutes in the OR
 * Alex completed a session of Knee rehab (2 sessions left).


```python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

class Patient(Person):
    def __init__(self, name, team, treatmen_plan):
        super().__init__(name)
        self.team = team
        self.treatment_plan = treatmen_plan
    
    def plan(self, dose=0, surg_time=0, therapy_amount=0):
            result = f'--- Care for {self.name} ---\n\n'
            for i in self.team:
                result += f'{self.name} {i}\n' 
                
            for i in self.treatment_plan:
                if type(i) == Medication:
                    i - dose
                    result += f'{self.name} was given {dose}mg of {i}\n'
                elif type(i) == Surgery:
                    i - surg_time
                    result += f'{self.name} {i}\n'
                elif type(i) == Therapy:
                    i - therapy_amount
                    result += f'{self.name} completed {therapy_amount} session\s {i}\n'
            return result

class Care_provider(Person):
    def __init__(self, name):
        super().__init__(name)
        


class Care_team:
    def __init__(self, team):
        self.team = team

    def add_member(self,member):
        self.team.append(member)

    def __str__(self):
        result = ""
        for i in self.team:
            result += f'{i}, '
        return result

    def __iter__(self):
        return iter(self.team)
        

class Doctor(Care_provider):
    def __init__(self, name):
        super().__init__(name)

    def __str__(self):
        return f'was examined and diagnosed by Dr {self.name}'


class Nurse(Care_provider):
    def __init__(self, name):
        super().__init__(name)

    def __str__(self):
        return f'had vitals checked by Nurse {self.name}'


class Specialist(Care_provider):
    def __init__(self, name, specialty):
        super().__init__(name)
        self.specialty = specialty

    def __str__(self):
        return f"had a {self.specialty} consult with Dr {self.name}"

class Treatment:
    def __init__(self, name, max_amount):
        self.name = name
        self.max_amount = max_amount
    def __str__(self):
        return f'{self.name}'

    def __sub__(self, amount):
        self.max_amount -= amount
        
class Treatment_plan:
    def __init__(self, treatments):
        self.treatments = treatments

    def add_treatment(self, treatment):
        self.treatments.append(treatment)

    def __str__(self):
        result = ""
        for i in self.treatments:
            result += f'{i}, '
        return result

    def __iter__(self):
        return iter(self.treatments)
        
class Medication(Treatment):
    def __init__(self, name, dose):
        super().__init__(name, dose)

    def __str__(self):
        return f'{self.name} ({self.max_amount}mg remaining in stock)'

class Surgery(Treatment):
    def __init__(self, name, duration):
        super().__init__(name, duration)

    def __str__(self):
        return f'underwent {self.name} surgery, booked for {self.max_amount} minutes in the OR'

class Therapy(Treatment):
    def __init__(self, name, session_limit):
        super().__init__(name, session_limit)

    def __str__(self):
        return f'of {self.name} rehab ({self.max_amount} sessions left).'

kim = Doctor('Kim')

rios = Nurse('Rios')

chen = Specialist('chen', 'cardiology')

team = Care_team([])

team.add_member(kim)

team.add_member(rios)
team.add_member(chen)

amoxcilin = Medication('Amoxcilin', 1200)
appendectomy = Surgery('Appendectomy', 90)
knee = Therapy('Knee', 3)


treatment_plan = Treatment_plan([])

treatment_plan.add_treatment(amoxcilin)
treatment_plan.add_treatment(appendectomy)
treatment_plan.add_treatment(knee)

alex = Patient('Alex', team, treatment_plan)

print(alex.plan(500, 0, 1))
```
