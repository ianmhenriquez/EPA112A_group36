### 2. Now let's make a nice poem!
A verse will be generated depending on a number n, which is the sum of the outcomes of throwing 10 dice. These are regular dice.  
The verses are generated by picking random words of certain parts-of-speech, as follows:

* If n is divisible by 7, the verse is 'And so the ' followed by an adjective, 'and ', another adjective, a noun and a verb

* If n is divisible by 3, the verse is composed of 3 adjectives (an 'and' before the last adjective), followed by a noun, an adverb, and a verb.

* If n is divisible by 2, the verse is the word 'The ' followed by 3 adjectives (an 'and' before the last adjective), followed by a noun and a verb.

* Otherwise the verse is 'Here and there and so it goes...'

Make your poem look pretty by adding punctuation marks, make a new verse start with a capital letter and add indentation for every second verse. The last verse should end with a dot.

**Generate a poem with 8 verses.**

### 2.2 And now time for coding. Write your code bellow.

```python
from collections import defaultdict
import random as r

word_dic = dict()

def create_dic(name):
    fin = open(name)
    dic = defaultdict(list)
    for line in fin:
        word_list = line.split()
        for word in word_list:
            match word[-2]:
                case 'a':
                    dic['a'].append(word[:-3])
                case 'n':
                    dic['n'].append(word[:-3])
                case 'd':
                    dic['d'].append(word[:-3])
                case 'v':
                    dic['v'].append(word[:-3])
    return dic

def sum_dice(amount=10):
    sum_of_dice = 0
    for i in range(amount):
        sum_of_dice += r.randint(1, 6)
    return sum_of_dice

def rand_word(dic, word_type):
    word = ''
    match word_type:
        case 'a':
            tmp = dic["a"]
            word = r.choice(tmp)
        case 'n':
            tmp = dic["n"]
            word = r.choice(tmp)
        case 'd': 
            tmp = dic["d"]
            word = r.choice(tmp)
        case 'v':
            tmp = dic["v"]
            word = r.choice(tmp)
    return word


def create_verse(dic, n):
    if n % 7 == 0:
        return f"And so the {rand_word(dic, 'a')} and {rand_word(dic, 'a')} {rand_word(dic, 'n')} {rand_word(dic, 'v')}"
    elif n % 3 == 0:
        return f"{rand_word(dic, 'a').capitalize()}, {rand_word(dic, 'a')} and {rand_word(dic, 'a')} {rand_word(dic, 'n')} {rand_word(dic, 'd')} {rand_word(dic, 'v')}"
    elif n % 2 == 0:
        return f"The {rand_word(dic, 'a')}, {rand_word(dic, 'a')} and {rand_word(dic, 'a')} {rand_word(dic, 'n')} {rand_word(dic, 'v')}"
    else:
        return "Here and there and so it goes..."

def create_poem(dic):
    poem = ''
    for i in range(8):
        if (i % 2 != 0):
            poem += '\t'
        n = sum_dice()
        poem += create_verse(dic, n)
        if (i == 7):
            poem+= '.'
        poem += '\n'
    return poem
    
    
word_dic = create_dic('poetic_words.txt')

poem = create_poem(word_dic)
print(poem)

```
